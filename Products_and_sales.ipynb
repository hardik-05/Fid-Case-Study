```python
import pandas as pd
```

```python
products=pd.read_csv('VA-Products.csv')
```

```python
products.head()
```

```python
print("Missing values:",products.isnull().sum())
```

```python
print("Unique values:",products.nunique())
```

```python
print("Distribution based on Segment")
print(products['Segement'].value_counts())
print("Distribution based on Category")
print(products['Category'].value_counts())
print("Distribution based on Manufacturer")
print(products['ManufacturerID'].value_counts())
```

```python
sales=pd.read_csv('VA-Sales.csv')
```

```python
sales.head()
```

```python
print("Missing Values in the sales table:\n",sales.isnull().sum())
```

```python
sales['Date'] = pd.to_datetime(sales['Date'])
sales['Year'] = sales['Date'].dt.year
print("Count of sales in each year: ")
print(sales.groupby('Year').size())
```

```python
sales['Month'] = sales['Date'].dt.month
print("Count of sales in each month: ")
print(sales.groupby('Month').size())
```

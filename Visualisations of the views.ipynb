```
import matplotlib.pyplot as plt
import numpy as np
my_df = cell20.to_pandas()
segments = my_df['SEGMENT_ID']
min_values = my_df['AVERAGE_ASSETS']
max_values = my_df['MAX_ASSETS']
avg_values =my_df['MIN_ASSETS']

# Bar width
bar_width = 0.25

# Set position of bar on X axis
r1 = np.arange(len(segments))
r2 = [x + bar_width for x in r1]
r3 = [x + bar_width for x in r2]

# Create bars
plt.bar(r1, min_values, width=bar_width, label='Min', color='blue')
plt.bar(r2, max_values, width=bar_width, label='Max', color='red')
plt.bar(r3, avg_values, width=bar_width, label='Avg', color='green')

# Add labels
plt.xlabel('Segments', fontweight='bold')
plt.ylabel('Asset Values', fontweight='bold')
plt.xticks([r + bar_width for r in range(len(segments))], segments)

# Add title and legend
plt.title('Min, Max, and Avg Asset Values per Segment')
plt.legend()

# Show the plot
plt.tight_layout()
plt.show()
```

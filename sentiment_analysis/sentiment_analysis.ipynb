```python
import pandas as pd
from textblob import TextBlob
import re
from datetime import datetime
```

```python
def clean_text(text):
    text = re.sub(r'[^a-zA-Z\s]','',text)
    text = text.lower()
    return text
```

```python
def get_sentiment(text):
    return TextBlob(text).sentiment.polarity
```

```python
def process_date (data_str) :
    try:
        return pd.to_datetime(data_str).strftime("%y-%m-%d")
    except:
        data_str
    
```

```python
csv_file_path = r'C:\Users\Administrator\a751809\Fid-Case-Study\sentiment_analysis\RedditNews.csv'
df = pd.read_csv(csv_file_path)

df.columns = ['date','news_text']

df['cleaned_text'] = df['news_text'].apply(clean_text)

df['sentiment_score'] = df[clean_text].apply(get_sentiment)

df['date'] = df['date'].apply(process_date)

df.to_csv(r"C:\Users\Administrator\a751809\Fid-Case-Study\sentiment_analysis\Sentiment_Score_RedditNews.csv")

print(f"Sentiment analysis done")
```

```python
def get_score_nltk(text):
    nltk.download('vader_lexicon',quiet=True)
    sia = SentimentIntensityAnalyzer()
    sentiment_score = sia.polarity_scores(text)
    return sentiment_score["compound"]
```

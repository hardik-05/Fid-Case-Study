```python
averaged_score_df = score_df.groupby('Date')['Sentiment_score'].mean().reset_index()
```

```python
merged_df  = pd.merge(averaged_score_df,index_df,on='Date',how = 'inner')
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

```python
fig = make_subplots(specs=[[{"secondary_y":True}]])
fig.add_trace(go.Candlestick(x=merged_df['Date'],
        open = merged_df['Open'],
        high = merged_df['High'],
        low = merged_df['Low'], 
        close = merged_df['Close'],
        name = "Price"),secondary_y = False)
        
```

```python
fig.add_trace(go.Scatter(x=merged_df['Date'],y=merged_df['Sentiment_score'],name="Sentiment Score",line=dict(color='black',width=1)),secondary_y = True)
```

```python
fig = go.Figure(data = go.Candlestick(x=merged_df['Date'],open = merged_df['Open'], high = merged_df['High'], low = merged_df['Low'], close = merged_df['Close']))

```

```python
fig.update_layout(title = 'Interactive Chart for Index Price and Sentiment Score',yaxis_title = 'Price',xaxis_title='Date',xaxis_rangeslider_visible = True ,template = 'plotly_dark')

```

```python
trace1 = go.Scatter(x=merged_df['Date'],y=merged_df['Close'], mode = 'lines+markers',name = 'Index Price',line = dict(color='blue'))
trace2 = go.Scatter(x=merged_df['Date'],y=merged_df['Sentiment_score'], mode = 'lines+markers',name = 'Sentiment Score',line = dict(color='green'))

layout = go.Layout(title = 'Index and sentiment score graph',xaxis = dict(title='Date'),yaxis = dict(title='Value'),hovermode = 'closest')

fig = go.Figure(data = [trace1],layout = layout)

fig.show()
 
```

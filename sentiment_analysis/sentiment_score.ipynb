```python
data.dropna(inplace=True)
```

```python
nlp = spacy.load("en_core_web_lg") 
def preprocess(text):
    doc = nlp(text)
    filtered_tokens = []
    for token in doc:
        if token.is_stop or token.is_punct:
            continue
        filtered_tokens.append(token.lemma_)
    #print("preprocess successful")
    return " ".join(filtered_tokens) 
```

```python
result = preprocess("Hello, you are a bad guy?? why are you not working on your grades!!!!")
print(result)
```

```python
import time
```

```python
start_time = time.time()
print(start_time)
data['Preprocessed Text'] = data['News'].apply(preprocess) 
end_time = time.time()
total_time_taken = end_time - start_time
print(total_time_taken)
```

```python
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
```

```python
def get_score_nltk(text):
    nltk.download('vader_lexicon',quiet=True)
    sia = SentimentIntensityAnalyzer()
    sentiment_score = sia.polarity_scores(text)
    return sentiment_score["compound"]
```

```python
data['Sentiment_score'] = data['Preprocessed Text'].apply(get_score_nltk) 
```
